```python
# Task 2: Data Visualization with Matplotlib & Seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load a real-world retail dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

# Display dataset information
print("DATASET OVERVIEW")
print(df.head())
print("\nDataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())

# Set visualization style
sns.set_theme(style="whitegrid")

# Create a figure containing five visualizations
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle("Retail Tips Dataset - Data Visualization Analysis",
             fontsize=18, fontweight="bold")

# 2. Bar Chart: Average bill by day
sns.barplot(data=df, x="day", y="total_bill",
            hue="day", legend=False, ax=axes[0, 0])
axes[0, 0].set_title("Average Bill by Day")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Average Bill")

# 3. Scatter Plot: Total bill vs tip
sns.scatterplot(data=df, x="total_bill", y="tip",
                hue="time", ax=axes[0, 1])
axes[0, 1].set_title("Total Bill vs Tip")
axes[0, 1].set_xlabel("Total Bill")
axes[0, 1].set_ylabel("Tip")

# 4. Histogram: Distribution of total bills
sns.histplot(data=df, x="total_bill", bins=20,
             kde=True, ax=axes[0, 2])
axes[0, 2].set_title("Distribution of Total Bills")
axes[0, 2].set_xlabel("Total Bill")
axes[0, 2].set_ylabel("Frequency")

# 5. Heatmap: Correlation between numerical variables
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True,
            cmap="coolwarm", fmt=".2f", ax=axes[1, 0])
axes[1, 0].set_title("Correlation Heatmap")

# 6. Box Plot: Total bill by day
sns.boxplot(data=df, x="day", y="total_bill",
            hue="day", legend=False, ax=axes[1, 1])
axes[1, 1].set_title("Bill Distribution by Day")
axes[1, 1].set_xlabel("Day")
axes[1, 1].set_ylabel("Total Bill")

# Hide the unused subplot
axes[1, 2].axis("off")

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig("data_visualization_analysis.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 7. Basic data analysis
print("\nKEY INSIGHTS")
print("Average total bill:", round(df["total_bill"].mean(), 2))
print("Average tip:", round(df["tip"].mean(), 2))
print("Highest total bill:", round(df["total_bill"].max(), 2))
print("Correlation between bill and tip:",
      round(df["total_bill"].corr(df["tip"]), 2))

print("\nAll five visualizations generated successfully!")
```
